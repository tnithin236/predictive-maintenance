# 03 Anomaly detection
Isolation Forest. Anomaly score is NOT a failure probability.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd, plotly.express as px
from src import config
from src.data_loader import load_raw_data
from src.preprocessing import rename_columns, clean_data
from src.preprocessing import split_data, get_xy
from src.anomaly_detection import build_isolation_forest, score_anomalies, summarize_against_labels
df, _ = clean_data(rename_columns(load_raw_data()))
tr, te = split_data(df); Xtr, _ = get_xy(tr); Xte, yte = get_xy(te)
iso = build_isolation_forest().fit(Xtr)
scored = score_anomalies(iso, Xte, te['machine_id'])
print(summarize_against_labels(scored, yte))
px.histogram(scored, x='anomaly_score', color='anomaly_status')
```
